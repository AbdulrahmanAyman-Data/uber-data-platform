```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("exploreation") \
    .master("local[*]") \
    .getOrCreate()
spark
```

```python
df = spark.read.parquet("hdfs://uber-hadoop-master:9000/data/bronze/rides/*")
```

```python
from pyspark.sql.functions import col

df2 = df

for column, dtype in df.dtypes:
    if dtype == "timestamp":
        df2 = df2.withColumn(column, col(column).cast("string"))

pdf = df2.limit(10).toPandas()

pdf
```
